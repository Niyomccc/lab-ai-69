## Build Recommendation Logic

### Subtask:
Develop a rule-based filtering function that receives user inputs and returns a set of recommended sports equipment.


**Reasoning**:
Create the recommendation logic function to filter and recommend sports equipment based on user inputs, followed by test cases to verify the logic.



```python
def recommend_equipment(df, sport=None, budget_category=None, skill_level=None, min_rating=None):
    """
    Recommends sports equipment from the DataFrame based on user criteria.

    Parameters:
    - df: pandas DataFrame of sports equipment
    - sport: str, the sport category (e.g., 'Tennis', 'Basketball', 'Running', 'Soccer')
    - budget_category: str, price category ('Low', 'Medium', 'High')
    - skill_level: str, user skill level ('Beginner', 'Intermediate', 'Advanced')
    - min_rating: float, minimum average rating (1.0 to 5.0)

    Returns:
    - pandas DataFrame with recommended products or a string message if no matches found.
    """
    filtered_df = df.copy()

    # Apply filters dynamically if parameters are specified
    if sport is not None:
        filtered_df = filtered_df[filtered_df['Sport'].str.lower() == sport.lower()]

    if budget_category is not None:
        filtered_df = filtered_df[filtered_df['Budget_Category'].str.lower() == budget_category.lower()]

    if skill_level is not None:
        filtered_df = filtered_df[filtered_df['Skill_Level'].str.lower() == skill_level.lower()]

    if min_rating is not None:
        filtered_df = filtered_df[filtered_df['Rating'] >= min_rating]

    # Check if we have any results remaining
    if filtered_df.empty:
        return "No matching items found. Try relaxing your filters."

    # Sort by Rating (descending) and Price (ascending / descending) to show best items
    filtered_df = filtered_df.sort_values(by=['Rating', 'Price'], ascending=[False, True])

    return filtered_df

# --- Test Scenarios ---
print("=== Test Scenario 1: Beginner Tennis Equipment ===")
test_1 = recommend_equipment(df_equipment, sport="Tennis", skill_level="Beginner")
print(test_1)

print("\n=== Test Scenario 2: High Budget Running gear with Rating >= 4.5 ===")
test_2 = recommend_equipment(df_equipment, sport="Running", budget_category="High", min_rating=4.5)
print(test_2)

print("\n=== Test Scenario 3: No Match Scenario ===")
test_3 = recommend_equipment(df_equipment, sport="Soccer", budget_category="Low", skill_level="Advanced")
print(test_3)
```

## Adapt Recommendation Logic for Thai

### Subtask:
Update the recommendation function to accept and filter using Thai criteria and test it with various queries.


**Reasoning**:
Update the recommendation logic function to support the Thai dataset and run test scenarios in Thai to verify correct functionality.



```python
def recommend_equipment_thai(df, sport=None, budget_category=None, skill_level=None, min_rating=None):
    """
    แนะนำอุปกรณ์กีฬาจาก DataFrame ตามเงื่อนไขที่ผู้ใช้ระบุ (รองรับภาษาไทย)

    Parameters:
    - df: pandas DataFrame ของอุปกรณ์กีฬาฉบับภาษาไทย
    - sport: str, ประเภทกีฬา เช่น 'เทนนิส', 'บาสเกตบอล', 'วิ่ง', 'ฟุตบอล'
    - budget_category: str, ระดับงบประมาณ เช่น 'ต่ำ', 'ปานกลาง', 'สูง'
    - skill_level: str, ระดับทักษะผู้ใช้ เช่น 'ผู้เริ่มต้น', 'ระดับกลาง', 'ระดับสูง'
    - min_rating: float, คะแนนรีวิวขั้นต่ำ (1.0 ถึง 5.0)

    Returns:
    - pandas DataFrame ที่ผ่านการกรองข้อมูล หรือข้อความภาษาไทยเมื่อไม่พบสินค้าที่ตรงเงื่อนไข
    """
    filtered_df = df.copy()

    # ล้างช่องว่างซ้าย-ขวาเพื่อความแม่นยำในการจับคู่ข้อความ
    if sport is not None:
        filtered_df = filtered_df[filtered_df['Sport'].str.strip() == sport.strip()]

    if budget_category is not None:
        filtered_df = filtered_df[filtered_df['Budget_Category'].str.strip() == budget_category.strip()]

    if skill_level is not None:
        filtered_df = filtered_df[filtered_df['Skill_Level'].str.strip() == skill_level.strip()]

    if min_rating is not None:
        filtered_df = filtered_df[filtered_df['Rating'] >= min_rating]

    # หากไม่พบสินค้า
    if filtered_df.empty:
        return "ไม่พบอุปกรณ์ที่ตรงตามเงื่อนไข กรุณาลองปรับลดตัวกรองใหม่อีกครั้ง"

    # จัดเรียงลำดับตามคะแนนรีวิวจากสูงไปต่ำ และราคาจากต่ำไปสูง เพื่อให้ได้ของดีและคุ้มค่าที่สุด
    filtered_df = filtered_df.sort_values(by=['Rating', 'Price'], ascending=[False, True])

    return filtered_df

# --- ทดสอบการใช้งานฟังก์ชันแนะนำอุปกรณ์กีฬา (ภาษาไทย) ---
print("=== กรณีทดสอบที่ 1: อุปกรณ์เทนนิสสำหรับ 'ผู้เริ่มต้น' ===")
test_thai_1 = recommend_equipment_thai(df_equipment, sport="เทนนิส", skill_level="ผู้เริ่มต้น")
print(test_thai_1)

print("\n=== กรณีทดสอบที่ 2: อุปกรณ์วิ่ง งบประมาณ 'สูง' ที่มีเรตติ้ง >= 4.5 ===")
test_thai_2 = recommend_equipment_thai(df_equipment, sport="วิ่ง", budget_category="สูง", min_rating=4.5)
print(test_thai_2)

print("\n=== กรณีทดสอบที่ 3: กรณีไม่พบผลลัพธ์ (ฟุตบอล, งบต่ำ, ระดับสูง) ===")
test_thai_3 = recommend_equipment_thai(df_equipment, sport="ฟุตบอล", budget_category="ต่ำ", skill_level="ระดับสูง")
print(test_thai_3)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# 1. Setup Thai Font in Matplotlib for App Rendering
try:
    font_path = 'Kanit-Regular.ttf'
    fm.fontManager.addfont(font_path)
    plt.rcParams['font.family'] = 'Kanit'
    plt.rcParams['axes.unicode_minus'] = False
except Exception as e:
    pass

# 2. Localized Thai Dataset
thai_data = [
    # เทนนิส (Tennis)
    {101: ("ไม้เทนนิส Wilson Federer", "เทนนิส", "ผู้เริ่มต้น", 45.0, "ต่ำ", 4.2)},
    {102: ("ไม้เทนนิส Babolat Pure Drive", "เทนนิส", "ระดับสูง", 219.0, "สูง", 4.8)},
    {103: ("ไม้เทนนิส Head Ti.S6", "เทนนิส", "ระดับกลาง", 99.0, "ปานกลาง", 4.5)},
    {104: ("ลูกเทนนิส Penn Championship (แพ็ก 3 ลูก)", "เทนนิส", "ผู้เริ่มต้น", 4.99, "ต่ำ", 4.7)},
    # บาสเกตบอล (Basketball)
    {105: ("ลูกบาสเกตบอล Spalding TF-1000", "บาสเกตบอล", "ระดับสูง", 69.99, "ปานกลาง", 4.6)},
    {106: ("ลูกบาสเกตบอล Wilson NCAA Replica", "บาสเกตบอล", "ระดับกลาง", 29.99, "ต่ำ", 4.4)},
    {107: ("แป้นบาสเกตบอลปรับระดับได้ Lifetime", "บาสเกตบอล", "ระดับกลาง", 249.99, "สูง", 4.3)},
    {108: ("ถุงเท้าบาสเกตบอล Nike Elite", "บาสเกตบอล", "ผู้เริ่มต้น", 14.0, "ต่ำ", 4.5)},
    # วิ่ง (Running)
    {109: ("รองเท้าวิ่ง Nike Pegasus 40", "วิ่ง", "ระดับกลาง", 130.0, "ปานกลาง", 4.7)},
    {110: ("รองเท้าวิ่ง Asics Gel-Kayano 30", "วิ่ง", "ระดับสูง", 160.0, "สูง", 4.8)},
    {111: ("รองเท้าวิ่ง Decathlon Kalenji Run Active", "วิ่ง", "ผู้เริ่มต้น", 24.99, "ต่ำ", 4.0)},
    {112: ("นาฬิกา GPS Garmin Forerunner 55", "วิ่ง", "ระดับกลาง", 199.99, "สูง", 4.6)},
    # ฟุตบอล (Soccer)
    {113: ("ลูกฟุตบอล Adidas Al Rihla World Cup", "ฟุตบอล", "ระดับสูง", 160.0, "สูง", 4.9)},
    {114: ("รองเท้าสตั๊ด Nike Mercurial Vapor Academy", "ฟุตบอล", "ระดับกลาง", 85.0, "ปานกลาง", 4.3)},
    {115: ("รองเท้าสตั๊ด Puma Future Play FG/AG", "ฟุตบอล", "ผู้เริ่มต้น", 55.0, "ปานกลาง", 4.1)},
    {116: ("สนับแข้ง Adidas Tiro Club", "ฟุตบอล", "ผู้เริ่มต้น", 12.0, "ต่ำ", 4.2)},
    {117: ("ตาข่ายประตูฟุตบอล Kwik Goal", "ฟุตบอล", "ระดับสูง", 350.0, "สูง", 4.7)},
    {118: ("ลูกบาสเกตบอล Wilson Evolution Game", "บาสเกตบอล", "ระดับสูง", 79.95, "ปานกลาง", 4.8)},
    {119: ("กระเป๋าเทนนิส Babolat Team", "เทนนิส", "ระดับกลาง", 65.0, "ปานกลาง", 4.4)},
    {120: ("กางเกงวิ่งขาสั้น Under Armour", "วิ่ง", "ผู้เริ่มต้น", 28.0, "ต่ำ", 4.3)}
]

flat_thai_data = []
for item in thai_data:
    for prod_id, details in item.items():
        flat_thai_data.append({
            "Product_ID": prod_id,
            "Product_Name": details[0],
            "Sport": details[1],
            "Skill_Level": details[2],
            "Price": details[3],
            "Budget_Category": details[4],
            "Rating": details[5]
        })

df_equipment = pd.DataFrame(flat_thai_data)

# 3. Streamlit UI Layout
st.set_page_config(page_title="ระบบแนะนำอุปกรณ์กีฬาอัจฉริยะ", layout="wide")

st.title("🏆 ระบบแนะนำอุปกรณ์กีฬาอัจฉริยะ (Sports Equipment Recommendation System)")
st.markdown("**ค้นหาอุปกรณ์กีฬาที่ตอบโจทย์ความต้องการและระดับทักษะของคุณได้อย่างรวดเร็วและคุ้มค่าที่สุด**")
st.write("--- ")

# 4. Sidebar Filter Widgets
st.sidebar.header("⚙️ ตัวกรองข้อมูล (Filters)")

# Sport category
sport_options = ["ทั้งหมด"] + list(df_equipment['Sport'].unique())
selected_sport = st.sidebar.selectbox("⚽ เลือกประเภทกีฬา", sport_options)

# Budget category
budget_options = ["ทั้งหมด"] + list(df_equipment['Budget_Category'].unique())
selected_budget = st.sidebar.selectbox("💰 ระดับงบประมาณ", budget_options)

# Skill level
skill_options = ["ทั้งหมด"] + list(df_equipment['Skill_Level'].unique())
selected_skill = st.sidebar.selectbox("👟 ระดับทักษะผู้เล่น", skill_options)

# Price Range slider
min_price_val = float(df_equipment['Price'].min())
max_price_val = float(df_equipment['Price'].max())
price_range = st.sidebar.slider(
    "💸 ช่วงราคา (USD)",
    min_value=0.0,
    max_value=400.0,
    value=(0.0, 400.0)
)

# 5. Filter Recommendation Logic
filtered_df = df_equipment.copy()

if selected_sport != "ทั้งหมด":
    filtered_df = filtered_df[filtered_df['Sport'] == selected_sport]

if selected_budget != "ทั้งหมด":
    filtered_df = filtered_df[filtered_df['Budget_Category'] == selected_budget]

if selected_skill != "ทั้งหมด":
    filtered_df = filtered_df[filtered_df['Skill_Level'] == selected_skill]

filtered_df = filtered_df[
    (filtered_df['Price'] >= price_range[0]) &
    (filtered_df['Price'] <= price_range[1])
]

# Sort recommendations: highest rating first, then lowest price
filtered_df = filtered_df.sort_values(by=['Rating', 'Price'], ascending=[False, True])

# 6. Main Content Area
col1, col2 = st.columns([3, 2])

with col1:
    st.subheader("📦 รายการอุปกรณ์กีฬาที่แนะนำ")
    if filtered_df.empty:
        st.warning("⚠️ ไม่พบอุปกรณ์ที่ตรงตามเงื่อนไข กรุณาลองปรับลดตัวกรองใหม่อีกครั้ง")
    else:
        st.dataframe(filtered_df.rename(columns={
            "Product_ID": "รหัสสินค้า",
            "Product_Name": "ชื่อสินค้า",
            "Sport": "ประเภทกีฬา",
            "Skill_Level": "ระดับทักษะ",
            "Price": "ราคา (USD)",
            "Budget_Category": "งบประมาณ",
            "Rating": "คะแนนรีวิว"
        }), use_container_width=True, hide_index=True)

with col2:
    st.subheader("📊 การแจกแจงวิเคราะห์ความคุ้มค่า")
    if not filtered_df.empty:
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.scatterplot(
            ax=ax,
            data=filtered_df,
            x="Price",
            y="Rating",
            hue="Sport",
            style="Skill_Level",
            s=150,
            palette="Set1",
            alpha=0.9
        )
        ax.set_title("วิเคราะห์ความคุ้มค่า (ราคา vs คะแนนรีวิว)", fontsize=14, fontweight="bold", pad=15)
        ax.set_xlabel("ราคา (USD)", fontsize=11)
        ax.set_ylabel("คะแนนรีวิว (1.0 ถึง 5.0)", fontsize=11)
        ax.set_xlim(left=max(0, price_range[0] - 10), right=price_range[1] + 10)
        ax.set_ylim(0.5, 5.5)
        ax.grid(True, linestyle="--", alpha=0.5)
        st.pyplot(fig)
    else:
        st.info("ไม่มีข้อมูลเพื่อสร้างแผนภูมิการวิเคราะห์")
```

The previous code block failed because there was already a process running on port 8501, or Streamlit was not running correctly. I will search for and terminate any existing processes running on port 8501 or any previous streamlit processes, and restart the Streamlit server cleanly.



```python
import subprocess
import time
import psutil

# 1. Force kill any existing python/streamlit processes running on port 8501
print("กำลังค้นหาและเคลียร์โปรเซสเก่าที่ค้างอยู่บนพอร์ต 8501...")
for proc in psutil.process_iter(['pid', 'name', 'cmdline']):
    try:
        cmd = proc.info['cmdline']
        if cmd and any('streamlit' in arg for arg in cmd):
            print(f"ปิดโปรเซส Streamlit เก่า: PID {proc.info['pid']}")
            proc.terminate()
            proc.wait(timeout=3)
    except Exception:
        pass

time.sleep(2)

# 2. Restart the Streamlit server
print("กำลังเริ่มต้นเซิร์ฟเวอร์ Streamlit ใหม่อีกครั้ง...")
process = subprocess.Popen(['streamlit', 'run', 'app.py', '--server.port', '8501', '--server.headless', 'true'])

# 3. Wait for initialization
time.sleep(8)

# 4. Verify process status
if process.poll() is None:
    print(f"✓ เซิร์ฟเวอร์ Streamlit เริ่มทำงานสำเร็จแล้ว! PID: {process.pid}")
    print("พร้อมสำหรับการเชื่อมต่อแล้ว")
else:
    print("❌ เซิร์ฟเวอร์ปิดตัวลงกะทันหัน ตรวจสอบรายละเอียดพอร์ตหรือไฟล์ app.py")
```

## สร้างลิงก์เชื่อมต่อสาธารณะที่เสถียร (Establish Stable Tunnel Link)

### Subtask:
ติดตั้งและตั้งค่า localtunnel เพื่อรับลิงก์เชื่อมต่อภายนอก (Public Tunnel URL) สำหรับการเข้าถึงตัวเว็บแอปพลิเคชันแนะนำอุปกรณ์กีฬา


**Reasoning**:
Expose the running Streamlit port 8501 to a public URL using localtunnel and display it for testing.



```python
import subprocess
import time
import os
import re
import urllib.request

# 1. ปิด tunnel เก่า (localtunnel) ที่อาจค้างอยู่ เพื่อไม่ให้ชนกัน
subprocess.run(["pkill", "-f", "lt --port"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
subprocess.run(["pkill", "-f", "cloudflared"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# 2. รอให้ Streamlit พร้อมก่อน (กันกรณีเปิดลิงก์แล้วเจอหน้า error)
print("กำลังตรวจสอบว่า Streamlit พร้อมใช้งาน...")
ready = False
for _ in range(30):
    try:
        if urllib.request.urlopen("http://localhost:8501/_stcore/health", timeout=2).status == 200:
            ready = True
            break
    except Exception:
        time.sleep(1)
print("✓ Streamlit พร้อมแล้ว" if ready else "⚠️ Streamlit ยังไม่ตอบสนอง (รันเซลล์ที่เริ่มเซิร์ฟเวอร์ก่อน)")

# 3. ดาวน์โหลด cloudflared (ครั้งเดียว) - ไม่ต้องสมัครสมาชิก และไม่มีหน้ากรอกรหัสผ่าน
if not os.path.exists("cloudflared"):
    urllib.request.urlretrieve(
        "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",
        "cloudflared"
    )
    os.chmod("cloudflared", 0o755)

# 4. เปิด Quick Tunnel ไปที่พอร์ต 8501 (เก็บ log ลงไฟล์เพื่อดึงลิงก์)
print("กำลังสร้างลิงก์สาธารณะ...")
log_file = open("cloudflared.log", "w")
tunnel_process = subprocess.Popen(
    ["./cloudflared", "tunnel", "--url", "http://localhost:8501", "--no-autoupdate"],
    stdout=log_file, stderr=subprocess.STDOUT
)

# 5. รอจนได้ลิงก์ (ปกติใช้เวลา 3-10 วินาที)
tunnel_url = None
for _ in range(30):
    time.sleep(1)
    with open("cloudflared.log", "r") as f:
        match = re.search(r"https://[-a-z0-9]+\.trycloudflare\.com", f.read())
    if match:
        tunnel_url = match.group(0)
        break

print("\n=======================================================")
if tunnel_url:
    print(f"✓ ลิงก์สำหรับให้ทุกคนเข้าใช้งาน: {tunnel_url}")
    print("เปิดได้ทันทีโดยไม่ต้องใส่รหัสผ่าน (ส่งลิงก์ให้อาจารย์/เพื่อนได้เลย)")
else:
    print("❌ ไม่สามารถสร้างลิงก์ได้ ลองรันเซลล์นี้ใหม่อีกครั้ง")
print("=======================================================")
print("หมายเหตุ: ลิงก์จะใช้ได้ตราบที่ Colab ยังเปิดอยู่และเซลล์นี้ยังไม่ถูกหยุด")
```
